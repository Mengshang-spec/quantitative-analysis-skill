```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), ".."))
os.chdir(os.path.join(os.getcwd(), ".."))

import pandas as pd
import numpy as np
from config import *
from src.data_loader import load_data, get_file_info
from src.data_cleaner import clean_data
from src.analysis import *
from src.quantitative_analysis import *
from src.visualize import *
%matplotlib inline
```

```python
# 1. 加载数据
df_raw = load_data(RAW_DATA_PATH)
get_file_info(df_raw)
df_raw.head()
```

```python
# 2. 清洗
df = clean_data(df_raw, drop_dup=True, num_strategy="median", cat_strategy="mode")
df.head()
```

```python
# 3. 基础分析
summary_stats(df)
missing_report(df)
numeric, categorical, _ = column_type_report(df)
```

```python
# 4. 类别权重 —— 改成你想要的列名
cat_col = categorical[0] if categorical else None
if cat_col:
    weight_df = category_frequency(df, cat_col, weight_method="frequency")
    plot_weight_distribution(weight_df, cat_col, top_n=10)
```

```python
# 5. 权重矩阵 —— 修改行列即可
if len(categorical) >= 2:
    matrix = build_weight_matrix(df, categorical[0], categorical[1])
    norm = normalize_matrix(matrix, method="total")
    plot_weight_matrix_heatmap(norm, title="权重矩阵")
```

```python
# 6. 相关系数
corr = correlation_matrix(df)
if not corr.empty:
    plot_correlation_heatmap(corr)
```
